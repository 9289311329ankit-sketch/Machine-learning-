```python
"reverse a number"
num = int(input("enter a number:"))
rev = 0
while num > 0:
  digit = num % 10
  rev = rev * 10 +  digit
  num = num // 10
  print("reverse a number" , rev)
```

```python
num = int(input("Enter a number: "))
original = num
rev = 0

while num > 0:
    rev = rev * 10 + num % 10
    num = num // 10

if original == rev:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
"Check prime"
num = int(input("Enter a number: "))

if num < 2:
    print("Not prime")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```

```python
"Number pyramid"
rows = int(input("Enter number of rows: "))

for i in range(1, rows + 1):
    print(" " * (rows - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
#   Manage a list of favourite items
favourites = []

while True:
    print("\n1. Add  2. Remove  3. Show  4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        item = input("Enter item: ")
        favourites.append(item)
        print(item, "added")
    elif choice == "2":
        item = input("Enter item to remove: ")
        if item in favourites:
            favourites.remove(item)
            print(item, "removed")
        else:
            print("Item not found")
    elif choice == "3":
        if favourites:
            for i, item in enumerate(favourites, 1):
                print(i, item)
        else:
            print("List is empty")
    elif choice == "4":
        break
    else:
        print("Invalid choice")
```

```python
# store and report subject marks
marks = {}
n = int(input("How many subjects? "))

for i in range(n):
    subject = input("Subject name: ")
    marks[subject] = int(input("Marks in " + subject + ": "))

total = sum(marks.values())
percentage = total / n

print("\n--- Report ---")
for subject, m in marks.items():
    print(subject, ":", m)

print("Total:", total)
print("Percentage:", percentage, "%")
print("Highest:", max(marks, key=marks.get), "-", max(marks.values()))
print("Lowest:", min(marks, key=marks.get), "-", min(marks.values()))
```

```python
"A Bank class holding the balance in a private style attribute with deposit and withdraw methods. A Library class that stores books and can add and list them."
class Bank:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance      # private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited Rs.", amount)
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawn Rs.", amount)

    def get_balance(self):
        return self.__balance


account = Bank("Amit", 1000)
account.deposit(500)
account.withdraw(300)
account.withdraw(5000)
print("Current balance: Rs.", account.get_balance())
```

```python
# CSV with DictWriter and DictReader
import csv

students = [
    {"name": "Amit", "marks": 78},
    {"name": "Riya", "marks": 35},
    {"name": "Sneha", "marks": 92}
]

# write
with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "marks"])
    writer.writeheader()
    writer.writerows(students)

# read
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["name"], "-", row["marks"])
```

```python
"Write and read a plain text file one name per line; write a CSV with csv.DictWriter and read it back with csv.DictReader; write and read a JSON file."
names = ["Amit", "Riya", "Sneha", "Rahul"]

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks            # goes through the setter below

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Marks must be a number")
        if value < 0 or value > 100:
            raise ValueError("Marks must be between 0 and 100")
        self._marks = value

    def grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 40:
            return "D"
        else:
            return "F"


def save_students(students, filename):
    try:
        with open(filename, "w") as f:
            for s in students:
                f.write(f"{s.name},{s.marks}\n")
        print("Saved", len(students), "students to", filename)
    except OSError as e:
        print("Could not write file:", e)


def load_students(filename):
    students = []
    try:
        with open(filename, "r") as f:
            for line_no, line in enumerate(f, 1):
                try:
                    name, marks = line.strip().split(",")
                    students.append(Student(name, int(marks)))
                except ValueError as e:
                    print(f"Line {line_no} skipped: {e}")
    except FileNotFoundError:
        print("File not found:", filename)
    except OSError as e:
        print("Could not read file:", e)
    return students


# ---- main program ----
students = []
try:
    n = int(input("How many students? "))
except ValueError:
    print("Please enter a whole number")
    n = 0

for i in range(n):
    name = input("Student name: ")
    try:
        marks = int(input("Marks of " + name + ": "))
        students.append(Student(name, marks))
    except ValueError as e:
        print("Invalid input, student not added:", e)

save_students(students, "students.txt")

print("\n--- Report ---")
for s in load_students("students.txt"):
    print(s.name, s.marks, "Grade:", s.grade())
```
