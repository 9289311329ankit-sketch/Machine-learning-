```python

# Import required libraries

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# Load wine dataset

X, y = load_wine(return_X_y=True, as_frame=True)

print("Dataset loaded successfully!")
```

```python

# (a) Print dataset shape

print("Shape of X:", X.shape)

# Count of each class

print("\nCount of each class:")
print(y.value_counts().sort_index())
```

```python

# (b) Split dataset into training and testing

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Print shapes

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python

# (c) Apply StandardScaler

scaler = StandardScaler()

# Fit and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Only transform testing data
X_test_scaled = scaler.transform(X_test)

print("Training data scaled successfully!")
print("Testing data scaled successfully!")
```

```python

# (d) Train Logistic Regression model

model = LogisticRegression(max_iter=1000)

# Train model
model.fit(X_train_scaled, y_train)

# Predict test data
y_pred = model.predict(X_test_scaled)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Test Accuracy:", accuracy)

# Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python

# (a) Decision Tree Classifier

dt_model = DecisionTreeClassifier(random_state=42)

# Train the model on unscaled data
dt_model.fit(X_train, y_train)

# Predict test data
dt_pred = dt_model.predict(X_test)

# Calculate accuracy
dt_accuracy = accuracy_score(y_test, dt_pred)

print("Decision Tree Test Accuracy:", dt_accuracy)
```

```python
# (b) Train Random Forest Classifier

rf_model = RandomForestClassifier(random_state=42)

# Train the model on unscaled data
rf_model.fit(X_train, y_train)

# Predict test data
rf_pred = rf_model.predict(X_test)

# Calculate accuracy
rf_accuracy = accuracy_score(y_test, rf_pred)

print("Random Forest Test Accuracy:", rf_accuracy)
```

```python
# (c) Feature importance

import pandas as pd
importances = rf_model.feature_importances_

# Create feature importance table
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importances
})

# Sort in descending order
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Print top 3 features
print("Top 3 Important Features:")
print(feature_importance.head(3))
```

```python

# (a) KNN on unscaled data

knn_unscaled = KNeighborsClassifier(n_neighbors=5)

# Train model using original data
knn_unscaled.fit(X_train, y_train)

# Predict test data
knn_pred_unscaled = knn_unscaled.predict(X_test)

# Calculate accuracy
knn_accuracy_unscaled = accuracy_score(
    y_test, knn_pred_unscaled
)

print("KNN Unscaled Accuracy:", knn_accuracy_unscaled)
```

```python

# (b) KNN on scaled data

knn_scaled = KNeighborsClassifier(n_neighbors=5)

# Train model using scaled data
knn_scaled.fit(X_train_scaled, y_train)

# Predict using scaled test data
knn_pred_scaled = knn_scaled.predict(X_test_scaled)

# Calculate accuracy
knn_accuracy_scaled = accuracy_score(
    y_test, knn_pred_scaled
)

print("KNN Scaled Accuracy:", knn_accuracy_scaled)
```

```python

# (c) Compare both accuracy values

print("KNN Unscaled Accuracy:", knn_accuracy_unscaled)
print("KNN Scaled Accuracy:", knn_accuracy_scaled)
```

```python

# Q4 (a): Scale all features

from sklearn.preprocessing import StandardScaler

scaler_kmeans = StandardScaler()

X_scaled = scaler_kmeans.fit_transform(X)

print("Original X shape:", X.shape)
print("Scaled X shape:", X_scaled.shape)
```

```python

# Q4 (b): Apply K-Means clustering

from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    n_init=10,
    random_state=42
)

# Get cluster labels
cluster_labels = kmeans.fit_predict(X_scaled)

print("Cluster labels:")
print(cluster_labels)
```

```python

# Q4 (d): Attach cluster labels to DataFrame

# Make a copy of original dataset
df = X.copy()

# Add cluster column
df["cluster"] = cluster_labels

# Calculate mean of selected features
cluster_means = df.groupby("cluster")[
    ["alcohol", "proline", "color_intensity"]
].mean()

print("Mean values for each cluster:")
print(cluster_means)
```

```python

# Create students.csv file

csv_data = """name,marks,city
Asha,78,Bangalore
Ravi,,Chennai
Meena,91,Bangalore
Imran,64,Hyderabad
Divya,88,
"""

with open("students.csv", "w") as file:
    file.write(csv_data)

print("students.csv created successfully!")
```

```python

import csv
import json

print("Libraries imported successfully!")
```

```python

def read_marks(path):

    # (a) Open file safely using try and except
    try:
        with open(path, "r") as file:

            # (b) Read CSV file using DictReader
            reader = csv.DictReader(file)

            total_rows = 0
            marks_count = 0
            total_marks = 0

            # Read each row
            for row in reader:

                total_rows += 1

                # (c) Skip rows with empty marks
                if row["marks"] == "":
                    continue

                marks = int(row["marks"])

                marks_count += 1
                total_marks += marks

            # Calculate average
            if marks_count > 0:
                average_marks = total_marks / marks_count
            else:
                average_marks = 0

            # Create result dictionary
            result = {
                "total_rows": total_rows,
                "rows_with_marks": marks_count,
                "average_marks": average_marks
            }

            return result

    # Handle missing file
    except FileNotFoundError:
        print(f"File not found: {path}")
        return None
```

```python
import statistics

def read_marks_median(path):
    try:
        with open(path, "r") as file:
            reader = csv.DictReader(file)

            total_rows = 0
            marks_list = []

            for row in reader:
                total_rows += 1

                # Skip rows with empty marks
                if row["marks"] == "":
                    continue

                marks_list.append(int(row["marks"]))

            # Calculate median
            if len(marks_list) > 0:
                median_marks = statistics.median(marks_list)
            else:
                median_marks = 0

            result = {
                "total_rows": total_rows,
                "rows_with_marks": len(marks_list),
                "median_marks": median_marks
            }

            return result

    except FileNotFoundError:
        print(f"File not found: {path}")
        return None

# Test the new function
stats_median = read_marks_median("students.csv")
if stats_median is not None:
    print(json.dumps(stats_median, indent=4))
```

```python
# Call the function and print the result as a formatted JSON string
file_path = "students.csv"
stats = read_marks(file_path)

if stats is not None:
    print(json.dumps(stats, indent=4))
```
